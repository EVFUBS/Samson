```python
from numpy.random import RandomState
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Dense, LSTM, Dropout, Embedding
import pandas as pd
import tensorflow as tf

df = pd.read_csv(r'C:\Users\lssmith\Documents\pdrepos\Samson\SamsonConsoleApp\SamsonPy\Data\SamsonActions.csv')
df['class'] = df['class'].astype('category')
df['class_enc'] = df['class'].cat.codes

x_train, x_test, y_train, y_test = train_test_split(df['text'], df['class_enc'], test_size=.2, random_state=42)
```

```python
num_classes = df['class'].nunique()
vocab_size = 50
encoder = tf.keras.layers.TextVectorization(
    max_tokens=vocab_size,  # Set the maximum vocabulary size (or leave it as None)
    standardize="lower_and_strip_punctuation",  # Standardize text (lowercase and remove punctuation)
    split="whitespace",  # Split text into tokens based on whitespace
    output_mode="int",  # Output integer sequences
    output_sequence_length=None,  # Set the maximum sequence length (or leave it as None)
    pad_to_max_tokens=True,  # Pad sequences to the same length (if True)
    vocabulary=None  # Optionally provide a pre-defined vocabulary
)
encoder.adapt(df['text'].to_numpy().flatten().tolist())
```

```python
x_train = encoder(x_train)
y_train = tf.keras.utils.to_categorical(y_train, num_classes=num_classes)
x_test = encoder(x_test) 
y_test =tf.keras.utils.to_categorical(y_test, num_classes=num_classes)
```

```python
# will need to determine the structure of the model with testing prob need some LSTM or RNN
model = Sequential([
    Embedding(input_dim=len(encoder.get_vocabulary()),output_dim=64),
    LSTM(128, return_sequences=True, activation="relu"),
    Dropout(0.2),
    LSTM(64, activation='relu'),
    Dropout(0.2),
    Dense(num_classes, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
history = model.fit(x_train, y_train, epochs=30, validation_data=(x_test, y_test))
model.save("ActionModel")
```
